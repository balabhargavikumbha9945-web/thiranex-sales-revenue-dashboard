```python
import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as px
import ipywidgets as widgets
from IPython.display import display

print("Libraries imported successfully!")
```

```python
# Sales by Category
category_data = pd.DataFrame({
    "Category": [
        "Technology",
        "Office Supplies",
        "Other",
        "Furniture"
    ],
    "Sales": [
        1.30,
        0.53,
        0.33,
        0.09
    ]
})

# Sales by Month
monthly_data = pd.DataFrame({
    "Month": [
        "January",
        "February",
        "March",
        "April",
        "May",
        "June",
        "July"
    ],
    "Sales": [
        0.11,
        0.07,
        0.05,
        0.17,
        0.23,
        0.12,
        0.26
    ]
})

# Top 5 Products
product_data = pd.DataFrame({
    "Product": [
        "Power Bank Shell 5V",
        "HDMI Cable 2.0 4K",
        "Bamboo Folding Portable",
        "Transparent Face Mask",
        "Rovtop USB Gaming"
    ],
    "Sales": [
        0.34,
        0.32,
        0.27,
        0.22,
        0.14
    ]
})

# Top 5 Customers
customer_data = pd.DataFrame({
    "Customer": [
        "Xylona Preis",
        "Mona Khan",
        "Kim Larson",
        "Yoga Bobby",
        "Arianne Irving"
    ],
    "Sales": [
        0.27,
        0.20,
        0.15,
        0.13,
        0.10
    ]
})

# Profit percentage by segment
segment_data = pd.DataFrame({
    "Segment": [
        "Corporate",
        "Home Office",
        "Consumer"
    ],
    "Profit_Percentage": [
        13.44,
        19.12,
        67.44
    ]
})

print("Dashboard data created successfully!")
```

```python
print("SALES BY CATEGORY")
display(category_data)

print("SALES BY MONTH")
display(monthly_data)

print("TOP 5 PRODUCTS")
display(product_data)

print("TOP 5 CUSTOMERS")
display(customer_data)

print("PROFIT BY SEGMENT")
display(segment_data)
```

```python
total_customers = 116
total_orders = 171
total_quantity = 23000
total_sales = 2.25
total_profit = 1.01

print("========== KEY PERFORMANCE INDICATORS ==========")
print("Total Customers :", total_customers)
print("Total Orders    :", total_orders)
print("Total Quantity  :", total_quantity)
print("Total Sales     :", str(total_sales) + " Million")
print("Total Profit    :", str(total_profit) + " Million")
```

```python
fig = px.bar(
    category_data,
    x="Category",
    y="Sales",
    title="Sales by Category",
    text="Sales"
)

fig.update_traces(texttemplate="%{text:.2f}M", textposition="outside")

fig.show()
```

```python
fig = px.line(
    monthly_data,
    x="Month",
    y="Sales",
    markers=True,
    title="Monthly Sales Trend"
)

fig.update_traces(text=monthly_data["Sales"])

fig.show()
```

```python
fig = px.bar(
    product_data,
    x="Sales",
    y="Product",
    orientation="h",
    title="Top 5 Products by Sales",
    text="Sales"
)

fig.update_traces(texttemplate="%{text:.2f}M", textposition="outside")

fig.show()
```

```python
fig = px.bar(
    customer_data,
    x="Sales",
    y="Customer",
    orientation="h",
    title="Top 5 Customers by Sales",
    text="Sales"
)

fig.update_traces(texttemplate="%{text:.2f}M", textposition="outside")

fig.show()
```

```python
fig = px.pie(
    segment_data,
    names="Segment",
    values="Profit_Percentage",
    title="Profit Percentage by Segment",
    hole=0.4
)

fig.show()
```

```python
category_dropdown = widgets.Dropdown(
    options=["All"] + list(category_data["Category"]),
    description="Category:"
)

output = widgets.Output()

def filter_category(change):
    with output:
        output.clear_output()

        if change["new"] == "All":
            display(category_data)
        else:
            filtered = category_data[
                category_data["Category"] == change["new"]
            ]
            display(filtered)

category_dropdown.observe(filter_category, names="value")

display(category_dropdown)
display(output)

display(category_data)
```

```python
month_dropdown = widgets.Dropdown(
    options=["All"] + list(monthly_data["Month"]),
    description="Month:"
)

month_output = widgets.Output()

def filter_month(change):
    with month_output:
        month_output.clear_output()

        if change["new"] == "All":
            display(monthly_data)
        else:
            filtered = monthly_data[
                monthly_data["Month"] == change["new"]
            ]
            display(filtered)

month_dropdown.observe(filter_month, names="value")

display(month_dropdown)
display(month_output)

display(monthly_data)
```
